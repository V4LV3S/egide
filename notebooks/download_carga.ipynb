```python
import pandas as pd
import requests
from datetime import datetime
from dateutils import relativedelta

BASE_URL =  "https://apicarga.ons.org.br/prd/"
ENDPOINT = ["cargaverificada"]
CODAREA = ["SE"]

init_date = datetime(2023, 2, 1)
end_date = datetime(2026, 9, 1)
```

```python
for endpoint in ENDPOINT:
    for codarea in CODAREA:
        dataset = []
        for date in pd.date_range(init_date, end_date, freq="MS"):
            di = date - relativedelta(months=1)
            print(endpoint, codarea, di)

            url = (
                f"{BASE_URL}{endpoint}?dat_inicio={di.strftime('%Y-%m-%d')}"
                f"&dat_fim={date.strftime('%Y-%m-%d')}&cod_areacarga={codarea}"
            )
            
            response = requests.get(url)
            if response.status_code == 200:
                data = response.json()
                dataset.extend(data)
            else:
                print(f"Failed to fetch data for {endpoint} on {date.strftime('%Y-%m-%d')}")

        df = pd.DataFrame(dataset)
        # Optionally, save the DataFrame to a CSV file
        df.to_parquet(f"data/processed/carga/{codarea}_{endpoint}.parquet", index=False)
```

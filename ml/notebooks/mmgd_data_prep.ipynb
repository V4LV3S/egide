```python
import pandas as pd
```

```python
regioes = [
    'BA_SE', 'AL_PE', 'PB_RN','CE', 'MA', 'PI']

variaveis = [
    't2m', 'tcc', 'ssr', 'tp'
]

DIR_DATA = "../data"
DIR_METEO_DATA = f"{DIR_DATA}/meteoro/"
DIR_LOAD_DATA = f"{DIR_DATA}/mmgd/"
DIR_ML_DATA = f"{DIR_DATA}/training/"
```

```python
for regiao in regioes:
    ds_date = pd.read_parquet(f"{DIR_DATA}/date_features.parquet")
    ds_meteoro = pd.read_parquet(f"{DIR_METEO_DATA}/{regiao}/pca_features.parquet")
    ds_load = pd.read_parquet(f"{DIR_LOAD_DATA}/{regiao}_cargaverificada.parquet")

    ds_ml_input = (
        ds_meteoro
        .merge(ds_date, on="time", how="left", validate="one_to_one")
        .merge(
            ds_load.drop(columns="split"),
            on="time",
            how="left",
            validate="one_to_one",
        )
    )

    ds_ml_input.to_parquet(f"{DIR_ML_DATA}/{regiao}_ml_input.parquet")
```
